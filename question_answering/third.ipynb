## Q&A using transformers

```python
from transformers import pipeline
```

```python
context = """"
            The Great Wall of China is one of the most famous landmarks in the world. 
I           it stretches over 13,000 miles and was built to protect ancient Chinese states 
            from invasions. Construction began in the 7th century BC and continued for centuries. 
            Today, it is a UNESCO World Heritage site and a popular tourist destination. 
            Many visitors climb its steep steps to enjoy breathtaking views. 
            Legends say that the wall is visible from space, although this is a common myth. 
            It remains a symbol of Chinese strength and perseverance.
          """

questions = [
    'Why visitors climb the great wall of china',
    'how long is the wall of china',
    'What is the name of the UNESCO world heritage site explained in above paragraph',
    'What is a common myth',
    'When did the construction began and how long did it lasted', 
]
```

### model 1

```python
pipeline_1 = pipeline('question-answering', model = 'distilbert-base-uncased-distilled-squad')
```

```python
score_model_1 = []
```

```python
for q in questions :
    results = pipeline_1(question = q, context = context)

    print(f"Q : {q}")
    print(f"Ans : {results}")

    score_model_1.append(results['score'])
```

### model 2

```python
pipeline_2 = pipeline('question-answering', model = 'bert-large-uncased-whole-word-masking-finetuned-squad')
```

```python
score_model_2 = []
```

```python
for i in range(0, len(questions)) : 
    results = pipeline_2(question = questions[i], context = context)
    print(f"Q : {questions[i]}")
    print(f"Ans : {results}")

    score_model_2.append(results)
```
