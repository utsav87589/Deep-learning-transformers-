## Q&A play with models

```python
import pandas as pd
import numpy as np
from transformers import pipeline
```

### loading the context

```python
import os
path = '../text_summarization'
files = os.listdir(path)
print(files)
```

```python
df = pd.read_csv(path + '/news_summary.csv')
df.head()
```

```python
df.tail()
```

```python
context = df['ctext'][4510]
print(context)
```

### generating questions

```python
gen_qa = pipeline('text2text-generation', model = 'iarfmoose/t5-base-question-generator')
```

```python
i = 0

while i < 5:

    gen_questions = gen_qa(context, max_length = 32, num_return_sequences = 3, do_sample = True, top_k = 50, top_p = 0.95, temperature = 0.8)
    for q in gen_questions : 
        first, last = q['generated_text'].split("?", 1)
        print(first)

    i += 1
```
