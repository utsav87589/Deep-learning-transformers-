## Q&A play with models

```python
import pandas as pd
import numpy as np
from transformers import pipeline
```

### loading the text for the questions

```python
df = pd.read_csv('../text_summarization/news_summary.csv')
df.head()
```

```python
for i in range(0, 10) : 
    print(f"{len(df['ctext'][i])}")
```

```python
context = df['ctext'][7]
```

```python
context
```

### generating the questions

```python
qa_gen = pipeline('text2text-generation', model = 'iarfmoose/t5-base-question-generator')
```

```python
questions = []
```

```python
for num in range(0, 5) : 
    question = qa_gen(context, max_length = 24, do_sample = True, top_p = 0.95, num_return_sequences = 3)
    for i, q in enumerate(question) : 
        first, rest = q['generated_text'].split('?', 1)
        print(f"Q{i+1} : {first}")
```
