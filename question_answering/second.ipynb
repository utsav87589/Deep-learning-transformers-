## Q&A using transformers

### Revision using the 2 different models

```python
from transformers import pipeline
```

```python
context = """
          Australia won the 2023 world cup, this cricket world cup was hosted by India,
          India and Australia played the final in Ahemdabad, Narendra modi stadium,
          Australian skipper Pat Cummins won the toss and decided to bowl first, where 
          They managed to restrict India to a small total score of 240, and then Austrlia won the final in 
          a convinving victory with Travis head scoring 140, and by 6 wickets in hand, It was Australia's sixth world cup win,
          after 1987, 1999, 2003, 2007 and 2015, making the Australia number 1 ranked team in cricket.
          """

questions = [
    'Who hosted the 2023 Cricket world cup',
    'How many runs Travis Head score',
    'Which team won the toss',
    'Where did the both team played the final',
    'Which teams played the 2023 world cup final',
    'Which side won the 1999 world cup',
    'How many runs Indian team scored in the final'
]
```

### model 1

```python
qa_pipeline_1 = pipeline('question-answering', model = 'distilbert-base-uncased-distilled-squad')
```

```python
scores_1 = []
```

```python
for q in questions :
    results = qa_pipeline_1(question = q, context = context)
    print(f"Q : {q}")
    print(f"Ans : {results}")
    scores_1.append(results['score'])
```

### model : 2

```python
qa_pipeline_2 = pipeline('question-answering', model = 'bert-large-uncased-whole-word-masking-finetuned-squad')
```

```python
scores_2 = []
```

```python
for i in range(0, len(questions)) : 
    result = qa_pipeline_2(question = questions[i], context = context)
    print(f"Q : {questions[i]}")
    print(f"Ans : {result}")
    scores_2.append(result['score'])
```
