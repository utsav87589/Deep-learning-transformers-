### loading the model


```python
from huggingface_hub import login
```

```python
qa_gen = pipeline('text-generation', model = 'google/vaultgemma-1b')
```

### Zero shot prompting

```python
# prompt = f"generate 5 different questions from the following text : {context} "
```

```python
prompt = f"""
            You are a question generator. 
Read the passage below and generate 5 unique, clear, and relevant questions that could be asked about it. 
Do not repeat the passage, only output the questions in a numbered list.

Passage:
{context}

Questions : 
1.
"""
```

```python
questions = qa_gen(prompt, max_new_tokens = 25, temperature = 0.8, do_sample = True)
```

```python
print(f"{type(questions)} :: {len(questions)}")
```
