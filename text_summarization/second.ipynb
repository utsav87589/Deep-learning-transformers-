## Text Summarization

```python
import pandas as pd
import numpy as np
from transformers import pipeline
```

### loading the data

```python
df = pd.read_csv('news_summary_more.csv')
df.head()
```

```python
df = df.head(10)
```

```python
df
```

```python
df['headlines'][0]
```

### model 1

```python
model_1 = pipeline('summarization', model = 'sshleifer/distilbart-cnn-12-6')
```

```python
gen_summary = []
for i in range(0, 10) : 
    summary = model_1(df['text'][i], min_length = 12, max_length = 15, do_sample = False)
    gen_summary.append(summary)

    print(f"text : {df['text'][i]}")
    print(f"ref summary : {df['headlines'][i]}")
    print(f"gen summary : {summary}")

    print('*'*50)
```

### model 2

```python
model_2 = pipeline('summarization', model = 'facebook/bart-large-cnn')
```

```python
gen_summary_2 = []
for i in range(0, 10) : 
    summary = model_2(df['text'][i], min_length = 12, max_length = 15, do_sample = False)
    gen_summary_2.append(summary)

    print(f"text : {df['text'][i]}")
    print(f"ref summary : {df['headlines'][i]}")
    print(f"gen summary : {summary}")

    print('*'*50)
```
