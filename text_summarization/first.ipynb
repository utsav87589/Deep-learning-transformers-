## Text summarization

```python
import pandas as pd
import numpy as np
from transformers import pipeline
```

### loading the data

```python
df = pd.read_csv('../text_summ_fine_tuning/Datasets/with_all_three/df_train.csv')
df.head()
```

```python
print(f"{len(df['headlines'][0])} :: {len(df['text'][0])} :: {len(df['ctext'][0])}")
```

```python
df = df.head(4)
```

### without ctext

```python
df
```

```python
df.drop(['Unnamed: 0', 'ctext'], axis = 1, inplace = True)
df
```

### model 1

```python
pipeline_1 = pipeline('summarization', model = 'sshleifer/distilbart-cnn-12-6')
```

```python
summary_1 = []
for i in range(0, 4) : 
    summary = pipeline_1(df['text'][i], max_length = 20, min_length = 10, do_sample = False)
    print(f"context : {df['text'][i]}")
    print(f"ref summary : {df['headlines'][i]}")
    print(f"generated summary : {summary}")

    print('*'* 40)


    summary_1.append(summary)
```

### model 2

```python
pipeline_2 = pipeline('summarization', model = 'facebook/bart-large-cnn')
```

```python
summary_2 = []
for i in range(0, 4) : 
    summary = pipeline_2(df['text'][i], min_length = 10, max_length = 20, do_sample = False)
    print(f"text : {df['text'][i]}")
    print(f"ref summary : {df['headlines'][i]}")
    print(f"gen summary : {summary}")

    print('*'* 40)

    summary_2.append(summary)
```
